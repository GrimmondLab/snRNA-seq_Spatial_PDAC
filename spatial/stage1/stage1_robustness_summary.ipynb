# Stage 1 robustness summary

Collects `cell_class_threshold` and Nhood results from the 16 sample notebooks. Run it after they have finished.

```python
import os

import pandas as pd

OUTPUT_ROOT = "/path/to/stage1_output"
SAMPLES = ["MH065_314", "MH074_314", "MH096_314", "MH097_314", "MH099_314", "MH104_314", "MH106_314", "MH107_314", "MH065_480", "MH074_480", "MH096_480", "MH097_480", "MH099_480", "MH104_480", "MH106_480", "MH107_480"]


def read_measures(path):
    return pd.read_csv(path, sep="\t", index_col="measure")["value"]


rows = []
for sample in SAMPLES:
    folder = os.path.join(OUTPUT_ROOT, sample)
    check = read_measures(os.path.join(folder, "stage1_check.tsv"))
    robustness_path = os.path.join(folder, "robustness_summary.tsv")
    has_test = os.path.exists(robustness_path)
    values = read_measures(robustness_path) if has_test else check
    rows.append({
        "sample": sample.split("_")[0],
        "panel": sample.split("_")[1],
        "same_as_saved_object": check["Same label"],
        "same_cell_class_threshold_label": values["Same label"],
        "nhood_z_correlation": values["Nhood z-score correlation"],
        "how_tested": "Recalculated PCA" if has_test else "Scoring only, no PCA used",
    })

table = pd.DataFrame(rows)
table.to_csv(os.path.join(OUTPUT_ROOT, "stage1_robustness_table.tsv"), sep="\t", index=False)
table
```
